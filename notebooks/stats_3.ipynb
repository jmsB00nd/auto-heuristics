```python
# %% [markdown]
# # Quantum Circuit Mapping Benchmarks
# This notebook processes benchmark data for swap counts and circuit depth across different datasets.

import pandas as pd

# %% [markdown]
# ## 1. Data Initialization
# We store the raw data from the document into dictionaries.

data = {
    "qasmbench_large": {
        "swap_count": {
            "Sabre": 577.77, "Cirq": 750.05, "pytket": 587.95, "qmap": 507.41, 
            "EOH": 532.41, "Funsearch": 524.5, "Reevo": 512.09, "mcts": 529.86
        },
        "depth": {
            "Sabre": 1066.91, "Cirq": 1153.32, "pytket": 1028.77, "qmap": 925.82, 
            "EOH": 920.32, "Funsearch": 926.32, "Reevo": 923.23, "mcts": 921.82
        }
    },
    "queko_bss_16qbt": {
        "swap_count": {
            "Sabre": 604.11, "Cirq": 934.96, "pytket": 1045.94, "qmap": 1265.18, 
            "EOH": 807.69, "Funsearch": 711.22, "Reevo": 714.96, "mcts": 595.39
        },
        "depth": {
            "Sabre": 981.52, "Cirq": 1978.33, "pytket": 1507.41, "qmap": 1420.7, 
            "EOH": 1344, "Funsearch": 1199.72, "Reevo": 1215.54, "mcts": 1009.01
        }
    },
    "queko_bss_54qbt": {
        "swap_count": {
            "Sabre": 6378.32, "Cirq": 10111.92, "qmap": 12426.19, "EOH": 6341.41,
            "Funsearch": 6480.32, "Reevo": 6180.44, "mcts": 5894.64
        },
        "depth": {
            "Sabre": 3398.66, "Cirq": 4565.43, "qmap": 3496.34, "EOH": 3008.03,
            "Funsearch": 3078.15, "Reevo": 2947.23, "mcts": 2986.02
        }
    },
    "queko_bss_81qbt": {
        "swap_count": {
            "Sabre": 14463.04, "Cirq": 18258.52, "qmap": 25539.5, "EOH": 13544.76,
            "Funsearch": 14919.39, "Reevo": 14259.1, "mcts": 13023.12
        },
        "depth": {
            "Sabre": 4984.08, "Cirq": 5244.8, "qmap": 4818.63, "EOH": 3958.21,
            "Funsearch": 5022.81, "Reevo": 4162.39, "mcts": 3868.66
        }
    }
}

# %% [markdown]
# ## 2. Improvement Calculation
# We define a function to calculate the % improvement relative to the 'Sabre' baseline.
# Positive values indicate a reduction (improvement) in swaps/depth.

def calculate_improvements(dataset_dict):
    results = {}
    for benchmark, metrics in dataset_dict.items():
        bench_results = {}
        for metric_name, values in metrics.items():
            baseline = values["Sabre"]
            # Calculate improvement: (Baseline - Value) / Baseline * 100
            improvements = {
                method: f"{((baseline - val) / baseline * 100):.2f}%" 
                for method, val in values.items()
            }
            
            # Combine raw values and improvement for a clear table
            combined = {
                method: {"Value": values[method], "Improvement": improvements[method]}
                for method in values
            }
            bench_results[metric_name] = pd.DataFrame(combined).T
            
        results[benchmark] = bench_results
    return results

processed_data = calculate_improvements(data)

# %% [markdown]
# ## 3. Results Display
# Showing the processed tables for each dataset.

for benchmark, tables in processed_data.items():
    print(f"\n{'='*30}\nBENCHMARK: {benchmark.upper()}\n{'='*30}")
    for metric, df in tables.items():
        print(f"\n--- {metric.replace('_', ' ').title()} ---")
        print(df)
```
